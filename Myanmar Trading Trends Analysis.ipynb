```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------
# Step 1: Filter top partners
# ---------------------------
df_topimports_2019 = df_2019[df_2019['Partner Name'].isin(['Singapore','Malaysia','Thailand','Indonesia','China'])]
df_topexports_2019 = df_2019[df_2019['Partner Name'].isin(['China','United States','Thailand','India','Japan'])]
df_topimports_2022 = df_2022[df_2022['Partner Name'].isin(['China','United States','Thailand','India','Japan'])]
df_topexports_2022 = df_2022[df_2022['Partner Name'].isin(['China','United States','Thailand','India','Japan'])]

# ---------------------------
# Step 2: Add a Year column
# ---------------------------
df_topimports_2019['Year'] = 2019
df_topexports_2019['Year'] = 2019
df_topimports_2022['Year'] = 2022
df_topexports_2022['Year'] = 2022

# ---------------------------
# Step 3: Keep relevant columns
# ---------------------------
df_topimports_2019 = df_topimports_2019[['Partner Name','Import (US$ Thousand)','Year']]
df_topexports_2019 = df_topexports_2019[['Partner Name','Export (US$ Thousand)','Year']]
df_topimports_2022 = df_topimports_2022[['Partner Name','Import (US$ Thousand)','Year']]
df_topexports_2022 = df_topexports_2022[['Partner Name','Export (US$ Thousand)','Year']]

# ---------------------------
# Step 4: Merge imports and exports
# ---------------------------
df_combined_2019 = pd.merge(df_topimports_2019, df_topexports_2019, on=['Partner Name','Year'], how='outer')
df_combined_2022 = pd.merge(df_topimports_2022, df_topexports_2022, on=['Partner Name','Year'], how='outer')

df_combined = pd.concat([df_combined_2019, df_combined_2022], ignore_index=True)

# ---------------------------
# Step 5: Reshape for plotting
# ---------------------------
df_melt = pd.melt(df_combined, id_vars=['Partner Name','Year'], 
                  value_vars=['Import (US$ Thousand)','Export (US$ Thousand)'],
                  var_name='Type', value_name='US$_Thousand')

# Clean Type column
df_melt['Type'] = df_melt['Type'].apply(lambda x: 'Import' if 'Import' in x else 'Export')

# ---------------------------
# Step 6: Plot combined chart
# ---------------------------
plt.figure(figsize=(12,6))
sns.barplot(data=df_melt, x='Partner Name', y='US$_Thousand', hue='Type', ci=None, palette='Set2')


plt.ylabel("US$ Thousand")
plt.title("Myanmar's Import and Export Earnings by Partner Country (2019 vs 2022)")
plt.xticks(rotation=45)
plt.legend(title='Type')
plt.tight_layout()
plt.show()
```
